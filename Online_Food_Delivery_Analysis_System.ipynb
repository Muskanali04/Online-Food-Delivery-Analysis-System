# DATA CLEANING

Explanation
What it does:

Why we use it:
These columns are extra index columns.

•	axis=1 → remove columns
•	inplace=True → permanently modify dataset
Example:

````python
# Remove unwanted columns
df.drop(['Unnamed: 0', 'Unnamed: 0.1'], axis=1, inplace=True)

# Rename columns for easy access
df.columns = [
    'restaurant_name',
    'restaurant_type',
    'rating',
    'num_ratings',
    'avg_cost',
    'online_order',
    'table_booking',
    'cuisines',
    'area',
    'address'
]

# Check updated columns
df.head()
````

# HANDLE MISSING VALUES

```text
df.dropna(subset=['rating', 'cuisines'], inplace=True)
Explanation
What it does:
Removes rows where:
•	rating is missing
•	cuisines are missing
Why we use it:
Incomplete values can affect analysis.
Example:
Before: | Rating | |—| | NaN |
After: Row removed.


````python
# Check missing values again
print(df.isnull().sum())

# Remove rows with missing ratings or cuisines
df.dropna(subset=['rating', 'cuisines'], inplace=True)

print("Missing values handled successfully")
````

# FIND BEST RESTAURANT BY CUISINE

```text
cuisine_name = input("Enter Cuisine Name: ")
Explanation
What it does:
Takes cuisine name from user.
Example:
Italian
________________________________________
result = df[df['cuisines'].str.contains(cuisine_name, case=False, na=False)]
Explanation
What it does:
Searches entered cuisine.
Meaning:
•	case=False → ignores capital/small letters
•	na=False → ignores missing values
________________________________________
result = result.sort_values(by='rating', ascending=False)
Explanation
What it does:
Sorts restaurants according to ratings.


````python
cuisine_name = input("Enter Cuisine Name: ")

result = df[df['cuisines'].str.contains(cuisine_name, case=False, na=False)]

result = result.sort_values(by='rating', ascending=False)

result[['restaurant_name', 'rating', 'avg_cost']].head(10)
````
